```python
# ===================== CONFIG =====================
ROOT_FOLDER = "/content/drive/MyDrive/Your-LIBRARY-folder"
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".tiff", ".bmp")

# ===================== SETUP ======================
from google.colab import drive
drive.mount('/content/drive')

!pip -q install pillow reportlab

import os
from PIL import Image
from reportlab.lib.pagesizes import A4, portrait, landscape
from reportlab.pdfgen import canvas

print("\n📌 ROOT FOLDER:", ROOT_FOLDER)
print("📂 Subfolders:", os.listdir(ROOT_FOLDER))

# ===================== IMAGE → A4 PDF ======================
def image_to_a4_pdf(image_path, pdf_path):
    img = Image.open(image_path).convert("RGB")
    w, h = img.size

    page_size = landscape(A4) if w > h else portrait(A4)
    page_w, page_h = page_size

    c = canvas.Canvas(pdf_path, pagesize=page_size)

    img_ratio = w / h
    page_ratio = page_w / page_h

    if img_ratio > page_ratio:
        new_w = page_w
        new_h = page_w / img_ratio
    else:
        new_h = page_h
        new_w = page_h * img_ratio

    x = (page_w - new_w) / 2
    y = (page_h - new_h) / 2

    c.drawImage(image_path, x, y, new_w, new_h)
    c.showPage()
    c.save()

# ===================== SCAN & CONVERT ======================
converted = 0
found = 0

for root, dirs, files in os.walk(ROOT_FOLDER):
    print(f"\n📂 Scanning: {root}")

    for file in files:
        print("   🔍 Found:", file)

        if file.lower().endswith(IMAGE_EXTENSIONS):
            found += 1
            img_path = os.path.join(root, file)
            pdf_path = os.path.join(root, os.path.splitext(file)[0] + ".pdf")

            if os.path.exists(pdf_path):
                print("   ⚠️ PDF exists, skipping")
                continue

            try:
                image_to_a4_pdf(img_path, pdf_path)
                converted += 1
                print("   ✅ Converted to PDF")
            except Exception as e:
                print("   ❌ Error:", e)

# ===================== SUMMARY ======================
print("\n==================== SUMMARY ====================")
print("🖼 Total images found:", found)
print("📄 PDFs created:", converted)
print("================================================")
```
